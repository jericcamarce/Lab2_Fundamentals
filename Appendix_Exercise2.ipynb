```python
SURNAME = "Camarce"
STUDENT_ID = "TUPM-26-1902"


print(f"System Initialized for: {SURNAME}")
print("--- GENERATING SIGNALS ---")

loops = 1
signals = ""

for i in range(0, loops + 6):
    SEED_NUM = STUDENT_ID[i]
    surname = SURNAME[i]
    print(f"{SEED_NUM}")
    print(f"{surname}")
    signals += SEED_NUM
    signals += surname

print("\n--- PROCESSING SIGNALS ---")
print(f"{signals}")

letters = 0
uppercase = 0
lowercase = 0
digits = 0
special_character = 0


print("\n--- ANALYZING SIGNAL CHARACTERS ---")
for i in range(0, loops + 6):
    SEED_NUM = STUDENT_ID[i]
    surname = SURNAME[i]
    
    if SEED_NUM.isalpha():
        letters += 1
        print(f"{SEED_NUM} - LETTER")
    elif SEED_NUM.isdigit():
        digits += 1
        print(f"{SEED_NUM} - DIGIT")
    else:
        special_character += 1
        print(f"{SEED_NUM} - SPECIAL CHARACTER")

    if surname.isalpha():
        letters += 1
        print(f"{surname} - LETTER")
    elif surname.isdigit():
        digits += 1
        print(f"{surname} - DIGIT")
    else:
        special_character += 1
        print(f"{surname} - SPECIAL CHARACTER")

print("\n--- SUMMARY AND FINAL CLASSIFICATION ---")
for i in range(0, loops + 6):
    SEED_NUM = STUDENT_ID[i]
    surname = SURNAME[i]
    
    if SEED_NUM.isupper():
        uppercase += 1
        print(f"{SEED_NUM} - UPPERCASE LETTER")
    elif SEED_NUM.islower():
        lowercase += 1
        print(f"{SEED_NUM} - LOWERCASE LETTER")
    elif SEED_NUM.isdigit():
        print(f"{SEED_NUM} - DIGIT")
    else:
        print(f"{SEED_NUM} - SPECIAL CHARACTER")

    if surname.isupper():
        print(f"{surname} - UPPERCASE LETTER")
        uppercase += 1
    elif surname.islower():
        print(f"{surname} - LOWERCASE LETTER")
        lowercase += 1
    elif surname.isdigit():
        print(f"{surname} - DIGIT")
    else:
        print(f"{surname} - SPECIAL CHARACTER")

print(f"\nSIGNAL: {signals}")
print(f"LETTERS: {letters}")
print(f"UPPERCASE LETTERS: {uppercase}")
print(f"LOWERCASE LETTERS {lowercase}")
print(f"DIGITS: {digits}")
print(f"SPECIAL CHARACTER: {special_character}")
```
