```python
SURNAME = "Camarce"
STUDENT_ID = "TUPM-26-1902"


print(f"System Initialized for: {SURNAME}")
print("--- GENERATING SENSOR READINGS ---")

loops = 1
valid_reading= 0
invalid_reading = 0
low_reading = 0
high_reading = 0

for i in range(0, loops + 6):
    SEED_NUM = STUDENT_ID[i]
    surname = SURNAME[i]
    reading = ord(surname) - ord(SEED_NUM)

    if reading < 0:
        invalid_reading += 1
        print( f"       {reading} - INVALID READING")
    else:
        valid_reading += 1
        print(f"        {reading} - VALID READING")

print("\n   --- CLASSIFYING READINGS ---")

loops = 1
for i in range(0, loops + 6):
    SEED_NUM = STUDENT_ID[i]
    surname = SURNAME[i]
    reading = ord(surname) - ord(SEED_NUM)

    if 0 < reading < 50:
        low_reading += 1
        print( f"        {reading} - LOW READING")
    elif reading > 50:
        high_reading += 1
        print(f"        {reading} - HIGH READING")

print("\n--- SUMMARY OF GENERATED DATA ---")
print(f"NO OF. DATA: {valid_reading + invalid_reading}")
print(f"VALID DATA: {valid_reading} ")
print(f"INVALID DATA: {invalid_reading} ")
print(f"HIGH READING: {high_reading} ")
print(f"LOW READING: {low_reading} ")


  


```
