```python
SURNAME = "Camarce"
STUDENT_ID = "TUPM-26-1902"

print(f"System Initialized for: {SURNAME}")
print("--- GENERATING SENSOR READINGS ---")

invalid_reading = 0
valid_reading = 0
low_reading = 0
high_reading = 0

loops = 1
for i in range(0, loops + 6):
    SEED_NUM = (STUDENT_ID[i])
    surname = SURNAME[i]
    readings = ord(surname) - ord(SEED_NUM)   

    if readings < 0:
        invalid_reading += 1
        print(f"    {readings}, INVALID READING")
    else:
        valid_reading += 1
        print(f"      {readings}, VALID READING")

print("\n--- CLASSIFYING READINGS ---")

for i in range(0, loops + 6):
    SEED_NUM = (STUDENT_ID[i])
    surname = SURNAME[i]
    readings = ord(surname) - ord(SEED_NUM)   
  
    if 0 < readings < 50:
        low_reading += 1
        print(f"      {readings}, LOW READING")
    elif readings > 50:
        high_reading += 1
        print(f"      {readings}, HIGH READING")

    
print("\n--- SUMMARY OF GENERATED DATA ---")
print(f"NO. OF DATA: {invalid_reading + valid_reading}")
print(f"VALID READING: {valid_reading}")
print(f"INVALID READING: {invalid_reading}")
print(f"LOW READING: {low_reading}")
print(f"HIGH READING: {high_reading}")

```
