```python
SURNAME = "Camarce"
STUDENT_ID = "TUPM-26-1902"

name_letter = SURNAME[-1]
letter_val = ord(name_letter)

SEED_NUM = int (STUDENT_ID[-1])

loops = SEED_NUM + 3

A = letter_val
B = SEED_NUM
C = loops


access_digits = ((A * B) - C) * 10
access_credential = (f"{SURNAME}_{access_digits}")

isLock = True

for i in range(1, loops + 1):
    while isLock:
        pwd = input("PASSWORD: ")
        if pwd == access_credential:
                isLock = False
                print("Unlocked.")
                break 
        else:
            print(f"REMAINING ATTEMPTS: {i}/{loops}")
            break
```
